```python
corpus = [
    
    "This is the first document.",
    "This document is the second document.",
    "And This is the third document.",
    "Is this the first document?",
    
]
print("Training Corpus:")
for doc in corpus:
    print(doc)
```

```python
#initialize vocabulary with unique characters from the corpus
unique_chars = set()
for doc in corpus:
    for char in doc:
        unique_chars.add(char)
        
vocab = list(unique_chars)
vocab.sort()
vocab        
```

```python
end_of_word="</w>"
vocab.append(end_of_word)
print("\n initial Vocabulary:")
print(vocab)
print(f"\n Vocabulary Size: {len(vocab)}")
```

```python
#pre-tokenize the corpus into words and then into characters
#we will split for simplicity and add the end of word token 

char_splits = {}
for doc in corpus:
    words = doc.split()
    for word in words:
        if word not in char_splits:
            char_list = list(word) + [end_of_word]
            #use touple for immutability if storing counts later - you cant change tuple once created (values or length)
            word_tuple = tuple(char_list)
            if word_tuple not in word_splits:
                word_splits[word_tuple] = 0
            word_splits[word_tuple] += 1 
print("\n Pre-tokenized word frequencies:")
print(word_splits)
                
```
